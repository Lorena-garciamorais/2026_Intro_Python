## **Exercise 5 : Composition**

Define a Catchment class that holds several Measurements objects keyed by name (composition). Add an add_series(name, series) method and a names() method returning the keys. Build one with two series.

## First attempt :




```python
class Catchment:
  def __init__(self, name, measurments) :
    self.name : list[str] = []
    self.measurments: list[float] = []

  def add_series(self, name, series) :
    self.measurments.append(series)

  def names(self):
    return self.measurments.keys()

  def __repr__(self) -> str:
    return f"Catchment({self.name}, {self.measurments})"


catchment1 = Catchment("serie1", [2, 4, 6, 8])
catchment2 = Catchment("serie2", [1, 2, 3, 4])
print(catchment1)
print(catchment2)
```

## Analysis of my mistakes :


1.   **Using a list insted of a dictionnary :** My first mistake was to create two empty lists, one for the names and another for the measurements, to store the information. I realized that an empty dictionary was more appropriate for storing the measurement series with their corresponding names, using key-value pairs where each name is a key associated with its measurement series as a value.
As a result, I also had to replace `.append()`, which is used to add elements to a list, by `self.measurements[name] = series`, which is used to add key-value pairs to a dictionary. I also had to modify `__repr__()` because I no longer needed `self.name`. Indeed, `self.measurements` already contains both the names (keys) and their corresponding measurement series (values).

2. **Incorrect use of .keys() :** I initially used `.keys()` on a list, which caused an error, because it's made for dictionary.

3. **Incorrect creation of Catchment object :** At first, I created two independent Catchment objects, one for each serie. However, the exercice involved to build one object with two series. To fix this, I had to create one object of the Catchment class `catchment = Catchment()`. Then, I used a `for` loop with `zip()` which allows to scroll through both lists simultaneously to match each name (key) with the correct set of measurements (value).



## Revised version :

```python
class Catchment:
  def __init__(self) :
    self.measurements = {}

  def add_series(self, name, series) :
    self.measurements[name] = series

  def names(self):
    return self.measurements.keys()

  def __repr__(self) -> str:
    return f"Catchment({self.measurements})"


catchment = Catchment()

series_names = ["serie1", "serie2"]
series_measurements = [[2, 4, 6, 8], [1, 2, 3, 4]]

for n, m in zip(series_names, series_measurements) :
  catchment.add_series(n, m)

print(catchment)
```
