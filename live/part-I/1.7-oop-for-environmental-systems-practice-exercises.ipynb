## Exercise 3: A dataclass record

Use `@dataclass` to define a `Reading` with `timestamp` (str) and `temp_celsius` (float). Show that two readings with equal fields compare equal, and print one to see the generated repr.

````python
# Your solution here
from dataclasses import dataclass

@dataclass
class Reading:
    timestamp: str
    temp_celsius: float

r = Reading("2024-06-01", 18.2)
print(r)                                                  # auto __repr__
print(r.temp_celsius)
print(Reading("2024-06-01", 18.2) == Reading("2024-06-01", 18.2))   # auto __eq__
````

## Exercise 4: Read and mutate methods, with validation

Define a `Measurements` class holding a list of floats, with `add(x)` to append and `mean()` to return the average. `mean()` should raise `ValueError` if there are no values. Demonstrate both.

````python
# Your solution here
class Measurements :
  def __init__(self, name:str) : # condition intiale qu'on choisit avant --> jungfraujoch = Measurements("Jungfraujoch")
    self.name = name
    self.measures_list : list[float] = []

  def add_measure(self, measure: float) -> None :
    self.measures_list.append(measure)

  def mean_measure(self) -> float :
    if not self.measures_list :
      raise ValueError(f"{self.name} has no readings")
    return sum(self.measures_list) / len(self.measures_list)
````

````python
jungfraujoch = Measurements("Jungfraujoch")
basel = Measurements("Basel")

for t in [-2.3, -1.1, 0.4]:
    jungfraujoch.add_measure(t)
for t in [18.0, 19.2, 17.5]:
    basel.add_measure(t)

print(jungfraujoch)                                  # __repr__ in action
print(basel)
print("Jungfraujoch mean :", round(jungfraujoch.mean_measure(), 2), "°C")
print("Basel mean :", round(basel.mean_measure(), 2), "°C")
````
